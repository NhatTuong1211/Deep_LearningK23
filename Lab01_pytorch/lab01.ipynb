## Bài 6. Phép toán và nhân ma trận

```python
a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\nMatrix multiplication:")
print(a @ b.T)
```

+ a * b là phép nhân từng phần tử tương ứng (element-wise) giữa hai Tensor.  

+ a @ b.T là phép nhân ma trận giữa Tensor a và ma trận chuyển vị của Tensor b.  

+ Khi a có kích thước 2 x 3 và b.T có kích thước 3 x 2, shape của kết quả a @ b.T là torch.Size([2, 2]).  

## Bài 7. Các hàm thống kê trên Tensor

```python
a = torch.randn(3, 4)

print(a)
print("\nMean:")
print(torch.mean(a, dim=1))

print("\nStandard deviation:")
print(torch.std(a, dim=1))

print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
```

+ Khi đặt dim=0, các hàm toán học sẽ tính toán thu gọn theo chiều dọc.

+ Khi thay bằng dim=1, các phép tính này sẽ được thực hiện theo chiều ngang.

+ Kết quả trả về sẽ có kích thước bằng số hàng của ma trận gốc.

## Bài 9. Gradient Descent tìm hệ số đa thức

```python
polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
     X * X,
     X,
     np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
```

+ optimizer.zero_grad(): Xóa các giá trị gradient đã tính từ epoch trước để tránh việc cộng dồn gradient. 

+ mse.backward(): Tính toán gradient của hàm mất mát theo các tham số dựa trên đồ thị tính toán.  

+ optimizer.step(): Cập nhật giá trị của các hệ số w dựa trên gradient vừa tính được.  

## Bài 10. Giải hệ phương trình bằng Autograd

```python
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

```python
print("y = 2x^2 - 3x + 5")

X_np = np.random.randn(20, 1) * 5
Y_np = 2 * (X_np**2) - 3 * X_np + 5
XX = np.hstack([
    X_np**2, 
    X_np, 
    np.ones_like(X_np)
])

w = torch.randn(3, 1, requires_grad=True)
x_tensor = torch.tensor(XX, dtype=torch.float32)
y_tensor = torch.tensor(Y_np, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.1)

print("\nInitial coefficients:")
print(w)

for epoch in range(1500):
    optimizer.zero_grad()
    y_pred = x_tensor @ w
    mse = torch.mean(torch.square(y_tensor - y_pred))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)
```
